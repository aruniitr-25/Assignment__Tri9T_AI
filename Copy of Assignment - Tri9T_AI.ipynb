```python
!unzip -q -o /content/data-20261021T155414Z-1-001.zip -d /content/
!pip install PyMuPDF
import re
import unicodedata
import hashlib
from typing import List, Dict, Any, Optional
import pymupdf  # PyMuPDF primary parser


class ASTNode:
    def __init__(
        self,
        node_id: str,
        number_label: Optional[str],
        title: str,
        level: int,
        page_start: int,
        page_end: int,
        parent_id: Optional[str] = None
    ):
        self.node_id = node_id
        self.number_label = number_label
        self.title = title
        self.level = level
        self.page_start = page_start
        self.page_end = page_end
        self.parent_id = parent_id
        self.children: List['ASTNode'] = []
        self.body_text: str = ""
        self.tables: List[List[Dict[str, str]]] = []
        self.ordered_lists: List[List[str]] = []
        self.cross_refs: List[str] = []
        self.warnings: List[str] = []
        self.content_hash: str = ""

    def compute_hash(self) -> str:
        """Computes deterministic hash over body text, structured tables, and ordered lists."""
        hasher = hashlib.sha256()
        hasher.update(self.body_text.encode('utf-8'))
        for table in self.tables:
            for row in table:
                serialized_row = "|".join(f"{k}:{v}" for k, v in sorted(row.items()))
                hasher.update(serialized_row.encode('utf-8'))
        for olist in self.ordered_lists:
            for item in olist:
                hasher.update(item.encode('utf-8'))
        self.content_hash = hasher.hexdigest()
        return self.content_hash

    def to_dict(self) -> Dict[str, Any]:
        return {
            "node_id": self.node_id,
            "number_label": self.number_label,
            "title": self.title,
            "level": self.level,
            "page_start": self.page_start,
            "page_end": self.page_end,
            "parent_id": self.parent_id,
            "body_text": self.body_text,
            "tables": self.tables,
            "ordered_lists": self.ordered_lists,
            "cross_refs": self.cross_refs,
            "warnings": self.warnings,
            "content_hash": self.content_hash,
            "children": [child.to_dict() for child in self.children]
        }


def normalize_text(text: str) -> str:
    """Fixes ligatures (ﬁ, ﬂ), maps non-breaking hyphens to '-', joins line breaks, collapses spaces."""
    if not text:
        return ""
    text = unicodedata.normalize("NFKC", text)
    text = text.replace("\u2011", "-").replace("\u2010", "-")
    text = re.sub(r'(\w+)-\n(\w+)', r'\1\2', text)
    text = re.sub(r'\s+', ' ', text)
    return text.strip()


class CT200PDFParser:
    def __init__(self, pdf_path: str):
        self.doc = pymupdf.open(pdf_path)
        self.nodes: List[ASTNode] = []
        self.root: Optional[ASTNode] = None

    def assert_structural_preconditions(self):
        """Asserts absence of headers, footers, page numbers, images, or multi-columns."""
        for page_num in range(len(self.doc)):
            page = self.doc[page_num]
            images = page.get_images()
            assert len(images) == 0, f"Page {page_num+1} contains unexpected images."
            blocks = page.get_text("blocks")
            for b in blocks:
                y0, y1 = b[1], b[3]
                assert not (y1 < 25 or y0 > 575), f"Page {page_num+1} contains header/footer artifact at y=({y0:.1f}, {y1:.1f})."

    def parse(self) -> ASTNode:
        self.assert_structural_preconditions()
        extracted_lines = []

        for page_idx, page in enumerate(self.doc):
            page_num = page_idx + 1
            blocks = page.get_text("dict")["blocks"]
            for b in blocks:
                if b.get("type") != 0:
                    continue
                for l in b["lines"]:
                    line_text = "".join(span["text" ] for span in l["spans"])
                    is_bold = any("Bold" in span["font"] or "bold" in span["font"] for span in l["spans"])
                    font_sizes = [span["size"] for span in l["spans"]]
                    avg_size = sum(font_sizes) / len(font_sizes) if font_sizes else 0.0
                    norm_line = normalize_text(line_text)
                    if norm_line:
                        extracted_lines.append({
                            "text": norm_line,
                            "is_bold": is_bold,
                            "size": round(avg_size, 1),
                            "x0": round(l["bbox"][0], 1),
                            "y0": round(l["bbox"][1], 1),
                            "page": page_num
                        })

        # Edge Case 10: Title Block Parsing (3 wrapped 22pt lines)
        title_lines = [l for l in extracted_lines if l["size"] >= 20.0 and l["is_bold"]]
        title_text = " ".join([l["text"] for l in title_lines])

        self.root = ASTNode(
            node_id="root",
            number_label=None,
            title=title_text,
            level=0,
            page_start=1,
            page_end=len(self.doc)
        )
        self.nodes.append(self.root)

        heading_re = re.compile(r'^(\d+(?:\.\d+)*)\.?\s+(.*)$')
        current_node = self.root
        node_stack: List[ASTNode] = [self.root]

        i = 0
        n_lines = len(extracted_lines)

        while i < n_lines:
            line = extracted_lines[i]

            if line["size"] >= 20.0:
                i += 1
                continue

            match = heading_re.match(line["text"])
            # Distinguish headings from indented list items (x0=50.8, non-bold)
            is_indented_list = (line["x0"] > 45.0 and not line["is_bold"])

            if match and line["is_bold"] and not is_indented_list:
                num_label = match.group(1)
                heading_title = match.group(2)
                depth = len(num_label.split('.'))

                warning_msg = None
                if depth == 1:
                    target_level = 1
                    parent_node = self.root
                else:
                    expected_parent_label = ".".join(num_label.split('.')[:-1])
                    parent_candidate = next((n for n in reversed(self.nodes) if n.number_label == expected_parent_label), None)

                    if parent_candidate:
                        parent_node = parent_candidate
                        target_level = parent_node.level + 1
                    else:
                        # Edge Case 1: Skipped Level (2.1.1.1 following 2.1)
                        nearest_ancestor = node_stack[-1]
                        parent_node = nearest_ancestor
                        target_level = parent_node.level + 1
                        warning_msg = f"numbering depth {depth} but nearest ancestor is {nearest_ancestor.number_label} (level {nearest_ancestor.level}); no {expected_parent_label} exists"

                # Edge Case 2: Out-of-order Sibling Warning
                if parent_node.children:
                    prev_num = parent_node.children[-1].number_label
                    if prev_num and num_label < prev_num:
                        seq_warn = "number_label out of sequence"
                        warning_msg = f"{warning_msg} | {seq_warn}" if warning_msg else seq_warn

                # Edge Case 3: Duplicate Heading Disambiguation
                node_id = f"node_{num_label}"
                if any(n.node_id == node_id for n in self.nodes):
                    node_id = f"{node_id}_p{parent_node.number_label}"

                new_node = ASTNode(
                    node_id=node_id,
                    number_label=num_label,
                    title=heading_title,
                    level=target_level,
                    page_start=line["page"],
                    page_end=line["page"],
                    parent_id=parent_node.node_id
                )

                if warning_msg:
                    new_node.warnings.append(warning_msg)

                parent_node.children.append(new_node)
                self.nodes.append(new_node)

                while len(node_stack) > target_level:
                    node_stack.pop()
                node_stack.append(new_node)
                current_node = new_node
                i += 1
                continue

            # Edge Case 4: Table-Only Section 2.1 & Multi-Line Row Table 4.2
            if line["is_bold"] and "Parameter" in line["text"] and "Value" in line["text"]:
                table_rows = []
                i += 1
                while i < n_lines and not extracted_lines[i]["is_bold"] and extracted_lines[i]["x0"] > 35.0:
                    full_txt = extracted_lines[i]["text"]
                    m = re.match(r'^(Measurement method|Pressure range|Pulse range|Accuracy \(pressure\)|Accuracy \(pulse\)|Power source|Display)\s+(.*)$', full_txt)
                    if m:
                        table_rows.append({"Parameter": m.group(1), "Value": m.group(2)})
                    i += 1
                current_node.tables.append(table_rows)
                continue

            if line["is_bold"] and "Code" in line["text"] and "Meaning" in line["text"]:
                table_rows = []
                i += 1
                curr_row = None
                while i < n_lines and not extracted_lines[i]["is_bold"] and extracted_lines[i]["x0"] > 35.0:
                    t_line = extracted_lines[i]
                    txt = t_line["text"]
                    x0 = t_line["x0"]

                    if x0 < 50.0 and re.match(r'^E\d+$', txt):
                        if curr_row:
                            table_rows.append(curr_row)
                        curr_row = {"Code": txt, "Meaning": "", "Device Behavior": ""}
                    elif curr_row:
                        if x0 < 150.0:
                            curr_row["Meaning"] = (curr_row["Meaning"] + " " + txt).strip()
                        else:
                            curr_row["Device Behavior"] = (curr_row["Device Behavior"] + " " + txt).strip()
                    i += 1
                if curr_row:
                    table_rows.append(curr_row)
                current_node.tables.append(table_rows)
                continue

            # Edge Case 5: Numbered List Processing in 3.3
            list_match = re.match(r'^(\d+)\.\s+(.*)$', line["text"])
            if list_match and line["x0"] > 45.0:
                list_items = []
                while i < n_lines:
                    l_curr = extracted_lines[i]
                    lm = re.match(r'^(\d+)\.\s+(.*)$', l_curr["text"])
                    if lm and l_curr["x0"] > 45.0:
                        list_items.append(l_curr["text"])
                    elif list_items and l_curr["x0"] > 55.0:
                        list_items[-1] = f"{list_items[-1]} {l_curr['text']}"
                    else:
                        break
                    i += 1
                current_node.ordered_lists.append(list_items)
                continue

            # Edge Case 11: Ensure Level-1 container nodes (1-8) keep empty body_text
            
if current_node.level == 1:
               
i += 1
               
continue

           
# Prose Paragraph Joining across Page Boundaries (Edge Case 7 & 8)
            
current_node.page_end = max(current_node.page_end, line["page"])

          
xrefs = re.findall(r'(?:see|Section)\s+([0-9\.\,\s]+)', line["text"], re.IGNORECASE)
            for ref in xrefs:
                refs_cleaned = [r.strip() for r in re.split(r'[,\s]+', ref) if re.match(r'^\d+\.\d+$', r.strip())]
                current_node.cross_refs.extend(refs_cleaned)

            if current_node.body_text and not current_node.body_text.endswith('.'):
                current_node.body_text += " " + line["text"]
            else:
                current_node.body_text = (current_node.body_text + " " + line["text"]).strip()

            i += 1

        for node in self.nodes:
            node.compute_hash()

        return self.root


# Defining root_v1 and root_v2 by parsing both manuals
parser_v1 = CT200PDFParser("data/ct200_manual.pdf")
root_v1 = parser_v1.parse()

parser_v2 = CT200PDFParser("data/ct200_manual_v2.pdf")
root_v2 = parser_v2.parse()
```

```python
import re
from typing import Dict, List, Any

class MockLLMClient:
    """Offline, deterministic LLM client for classification without network calls."""
    @staticmethod
    def classify_change(node_v1: Dict[str, Any], node_v2: Dict[str, Any]) -> str:
        if node_v1["content_hash"] == node_v2["content_hash"]:
            return "fresh"

        text_v1 = node_v1["body_text"]
        text_v2 = node_v2["body_text"]

        # Tokenizer tracking codes (E1-E6), numbers, and technical units
        param_pattern = re.compile(r'\b(?:E\d+|\d+(?:\.\d+)?(?:\s*(?:mmHg|bpm|cm|cycles|%|seconds|minutes|hours|years|DC|V))?)\b')
        tokens_v1 = set(param_pattern.findall(text_v1))
        tokens_v2 = set(param_pattern.findall(text_v2))

        table_tokens_v1 = set(param_pattern.findall(str(node_v1["tables"])))
        table_tokens_v2 = set(param_pattern.findall(str(node_v2["tables"])))

        if (tokens_v1 != tokens_v2) or (table_tokens_v1 != table_tokens_v2):
            return "material"

        return "stale_minor"


class DocumentMatcher:
    def __init__(self, root_v1: ASTNode, root_v2: ASTNode):
        self.map_v1 = {n.node_id: n.to_dict() for n in self._flatten(root_v1)}
        self.map_v2 = {n.node_id: n.to_dict() for n in self._flatten(root_v2)}

    def _flatten(self, node: ASTNode) -> List[ASTNode]:
        nodes = [node]
        for child in node.children:
            nodes.extend(self._flatten(child))
        return nodes

    def generate_diff(self) -> Dict[str, Any]:
        results = {}
        for node_id, n2 in self.map_v2.items():
            if node_id not in self.map_v1:
                results[node_id] = {"status": "added", "title": n2["title"]}
            else:
                n1 = self.map_v1[node_id]
                status = MockLLMClient.classify_change(n1, n2)
                results[node_id] = {"status": status, "title": n2["title"]}

        for node_id, n1 in self.map_v1.items():
            if node_id not in self.map_v2:
                results[node_id] = {"status": "removed", "title": n1["title"]}

        return results
```

```python
matcher = DocumentMatcher(root_v1, root_v2)
diff = matcher.generate_diff()

demo_selection = ["node_2.1.1.1", "node_3.2", "node_4.2", "node_5.1", "node_4.3"]
for node_id in demo_selection:
    print(f"Node: {node_id:<14} | Status: {diff[node_id]['status']:<10} | Title: {diff[node_id]['title']}")
```
